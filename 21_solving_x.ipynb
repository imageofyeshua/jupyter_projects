```python
import sympy as sym
from IPython.display import display,Math
```

```python
x = sym.symbols('x')

expr = 2*x + 4 - 9
sym.solve(expr)
```

```python
display(Math('\\text{The solution is } %s \\text{ is x = } %g' %(sym.latex(expr),sym.solve(expr)[0]) ))
```

```python
expr = x**2 - 4
sol = sym.solve(expr)

display(type(sol))
display(len(sol))
display(sol[0])

for i in range(0,len(sol)):
    print('Solution #' + str(i+1) + ' is ' + str(sol[i]))
```

```python
y = sym.symbols('y')

expr = x/4 - x*y + 5

display(sym.solve(expr,x))
display(sym.solve(expr,y))
```
