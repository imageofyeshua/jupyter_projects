```python
import sympy as sym
from IPython.display import display,Math
x = sym.symbols('x')

e1 = 2*x + x*(4-6*x) + x
e2 = -x * (2/x + 4/x**2) + (4+x)/(4*x)
e3 = (x+3)*(x-3)*x*(1/(9*x))

expr = [e1,e2,e3]

for expri in expr:
    display(Math('%s \\quad \\Longleftrightarrow \\quad %s' %(sym.latex(expri),sym.latex(sym.expand(expri)))))
```
