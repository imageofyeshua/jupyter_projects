```python
import sympy as sym
```

```python
number = 2 

fact_dict = sym.factorint(number)

print(type(fact_dict))
print(list(fact_dict.keys()))
print(list(fact_dict.values()))
print(len(fact_dict))
```

```python
for i in range(2,51):
    di = sym.factorint(i)
    ks = list(di.keys())
    if len(di) == 1 and di[ks[0]] == 1:
        print('%g is a prime number' %(i))
    else:
        print('%g is a composite number with prime factors %s' %(i,ks))
```
