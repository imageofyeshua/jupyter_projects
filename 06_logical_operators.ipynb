```python
x1 = 4
x2 = 6

if x1 > x2:
    print('%g is greater than %g' %(x1,x2))
elif x1 < x2:
    print('%g is less than %g' %(x1,x2))
else:
    print('%g is equal to %g' %(x1,x2))
```
