```python
# list of words
text = [ 'Our Father in heaven',
         'Sanctified be Thy Name',
         'Thy kingdom come',
         'Thy will be done on earth',
         'As it is in heaven',
         'Give us this day our daily break',
         'And Forgive us our debts',
         'As we also forgive our debtors',
         'Do not lead us into temptations',
         'But deliver us from the evil one',
         'For Thine is the kingdom',
         'And the power and the glory, forever. Amen' ]
text
```

```python
# separate into words by splitting by spaces
import re
re.split(r"\s+", text[0])
```

```python
# can recombine into a text
'-'.join( re.split(r'\s+',text[0]) )
```

```python
# also make lower-case
allwords = re.split(r'\s+',' '.join(text).lower())
allwords
```

```python
# find the unique words
voca = sorted(set(allwords))
voca
```

```python
print(f'{len(allwords)} words in the text, {len(voca)} words in the vocabulary')
```

```python
# the encoder is a python dictionary type
word2idx = {}
for i,word in enumerate(voca):
    word2idx[word] = i
word2idx
```

```python
# and a decoder
idx2word = {}
for i,word in enumerate(voca):
    idx2word[i] = word
idx2word
```

```python
print(f'The word "kingdom" has index {word2idx["kingdom"]}')
```

```python
print(f'The index "7" maps to the word : "{idx2word[7]}"')
```

```python
# select random words from the dictionary
import numpy as np
randidx = np.random.randint(0,len(voca),size=5)

# words of wisdom as a list of tokens
[ idx2word[i] for i in randidx ]
```

```python
# does it sound more wise as text??
' '.join( [idx2word[i] for i in randidx ])
```

```python
# translate the text into numbers
text_as_int = [ word2idx[word] for word in allwords ]

# and numbers back into text
for tokeni in text_as_int:
    print(f'Token {tokeni:2}: {idx2word[tokeni]}')
```
