```python
from IPython.display import display,Math
```

```python
for num in range(2,5):
    ans = (4*num + 3) < (17 - num**2)
    display(Math('x=%g: 4x + 3 < 17 - x^2 = %s' %(num,ans)))
```

```python
for num in range(2,5):
    ans = (8*num - 2) <= (-3*num + 42)
    display(Math('x=%g: 8x - 2 <= -3x + 42 = %s' %(num,ans)))
```
