```python
from numpy import random
```

```python
def guessTheNumber():
    num = random.randint(1,101)
    guess = int(input('Guess a number between 1 and 100'))

    while num != guess:
        if num > guess:
            print('Guess higher!')
        elif num < guess:
            print('Guess lower!')
        guess = int(input('Guess again: '))
                    
    print('Got it The right number was %g and your final guess was %g' %(num,guess))
```
