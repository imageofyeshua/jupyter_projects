```python
import sympy as sym
from IPython.display import display,Math
```

```python
x = sym.symbols('x')

p1 = 2*x**3 + x**2 - x
p2 = x**3 - x**4 - 4*x**2

display(Math('(%s) + (%s) = %s' %(sym.latex(p1),sym.latex(p2),sym.latex(p1+p2))))
```

```python
f1 = sym.Poly(x**2 + 2*x)
f2 = sym.Poly(-x**3 + 4*x)
f3 = sym.Poly(x**5 - x**4 + x/4 + 4)

polyList = [f1,f2,f3]

for poli in polyList:
    if poli.degree() % 2 == 0:
        print('The degree of %s is even, and the coefficients sum to %g.' %(poli.as_expr(),sum(poli.coeffs())))
    else:
        print('The degree of %s is odd, and there are %g coefficients.' %(poli.as_expr(),len(poli.coeffs())))
        
```

```python
x,y = sym.symbols('x,y')

fxy = 4*x**4 - 3*x**2 + x*y**2 - 9*y**3
gxy = -x**3 + 6*x**2*y + .8*y**3
```

```python
sym.expand(fxy*gxy).subs({x:5,y:-2})
```

```python
sym.expand(fxy).subs({x:5,y:-2})*sym.expand(gxy).subs({x:5,y:-2})
```

```python
display(Math('(%s) \\times (%s) = %s' %(sym.latex(fxy),sym.latex(gxy),sym.latex(sym.expand(fxy*gxy)))))
```

```python
xval = 5
yval = -2

fg = (fxy*gxy).subs({x:xval,y:yval})
print('Multiplied solution is %g' %fg)

fxy_ans = fxy.subs({x:xval,y:yval})
gxy_ans = gxy.subs({x:xval,y:yval})
print('Separate solution comes to %s' %(fxy_ans*gxy_ans))
```

```python
p5 = 4*x**5 - x
p6 = 2*x**3 - x

display(Math('\\frac{%s}{%s} = %s' %(sym.latex(p5),sym.latex(p6),sym.latex(p5/p6))))
display(Math('\\frac{%s}{%s} = %s' %(sym.latex(p5),sym.latex(p6),sym.latex(sym.expand(p5/p6)))))
display(Math('\\frac{%s}{%s} = %s' %(sym.latex(p5),sym.latex(p6),sym.latex(sym.simplify(p5/p6)))))
```

```python
x,y = sym.symbols('x,y')
pNum = x**6 + 2*x**4 + 6*x - y
pDen = x**3 + 3

for yi in range(5,16):
    tempnum = pNum.subs(y,yi)
    display(Math('%s = %s' %(sym.latex(tempnum/pDen),sym.latex(sym.simplify(tempnum/pDen)))))

    if sym.fraction(sym.simplify(tempnum/pDen))[1] == 1:
        rightanswer = yi

print('The answer satisfies our goal is y = %g' %rightanswer)
```

```python
f1 = x**2 + 4*x + 3
f2 = 2*y**2 - 1
f3 = 3*y**2 + 12*y

expr = [f1, f2, f3]

for ei in expr:
    strfact = str(sym.factor(ei))
    if strfact.find('(') != -1:
        display(Math('%s \\quad \\Rightarrow\\quad %s' %(sym.latex(ei),sym.latex(sym.factor(ei)))))
    else:
        display(Math('%s \\quad \\Rightarrow\\quad \\text{ not factorable!!}' %(sym.latex(ei))))
        
    
```
