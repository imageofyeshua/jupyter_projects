```python
nums = range(-5,6)
for num in nums:
    firstchar = ' '
    if num < 0:
        firstchar = ''
    if abs(num) % 2 == 0:
        print('%s%g is an even number' %(firstchar,num))
    else:
        print('%s%g is an odd number' %(firstchar,num))
```
