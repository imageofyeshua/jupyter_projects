```python
import sympy as sym
```

```python
# associative
from sympy.abc import x,y,z

expr1 = x*(4*y)
expr2 = (x*4)*y

expr1 - expr2 
```

```python
# commutative

e1 = 4*x*y
e2 = x*4*y
e3 = y*x*4
```

```python
print(e1.subs({x:3,y:2}))
print(e2.subs({x:3,y:2}))
print(e3.subs({x:3,y:2}))
```

```python
# distributive
a,b,c,d = sym.symbols('a,b,c,d')

expr = (a+b)*(c+d)
sym.expand(expr)
```

```python
from IPython.display import display,Math
x,y,z,w = sym.symbols('x,y,z,w')

x = w*(4-w) + 1/w**2 * (1+w)
f1 = x*(y+z)
f2 = 3/x + x**2

f1*f2

display(Math(sym.latex(f1*f2)))
display(Math(sym.latex(sym.sympify(f1*f2))))
display(Math(sym.latex(sym.sympify(f1*f2 - f2*f1))))
display(Math(sym.latex(sym.expand(f1*f2))))
```
