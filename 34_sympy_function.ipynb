```python
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display,Math
```

```python
# create a symbolic variable
beta = sym.var('beta')
xDomain = [-2,2]
# display(Math(sym.latex(beta)))
# function
x = np.linspace(xDomain[0],xDomain[1],41)
y = beta**2 + 3*beta**3 - beta**4

sym.plot(y, (beta, xDomain[0], xDomain[1]),xlabel='x',ylabel=None,
         title=f'$f(\\beta) = {sym.latex(y)}$')
```

```python
fx = sym.lambdify(beta,y)

g = fx(x)
plt.plot(x,g,label='$g=x^2+x^3-x^4$')
plt.grid()
plt.legend()
plt.xlim(xDomain[0],xDomain[1])
plt.xlabel('x')
plt.ylabel('y=f(x)')
plt.show()
```
