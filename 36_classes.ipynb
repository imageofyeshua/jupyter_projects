```python
# OOP (classes)

class Experiment(object):
    # constructor method
    def __init__(self,eType,dataN,status='ongoing'):
        self.expType = eType
        self.dataCount = dataN
        self.expStatus = status

    def increaseN(self,number):
        self.dataCount += number

    def changeExpStatus(self,status):
        self.expStatus = status

    def __str__(self):
        return 'Experiment type: ' + self.expType + ' has ' + str(self.dataCount) + ' as ' + self.expStatus
```

```python
myexperiment = Experiment('Brain',14500)
```

```python
print(myexperiment)
myexperiment.increaseN(10)
print(myexperiment)
myexperiment.changeExpStatus('stop')
print(myexperiment)
```
