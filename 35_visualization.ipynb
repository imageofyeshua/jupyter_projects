```python
import matplotlib.pyplot as plt
```

```python
x = 3
y = 5

plt.plot(x,y,'bs')

plt.axis('square')
plt.axis([-6,6,-6,6])
plt.grid()

plt.show()
```

```python
x = [-4,5,2,4,5,6,3]
y = [0,3,5,6,2,-3,4]

for i in range(0,len(x)):
    plt.plot(x[i],y[i],'o',label='point %s' %(1+i))

plt.axis('square')
plt.grid()
plt.legend()
plt.show()
```

```python
plt.plot(4,2,'r+')

axis = plt.gca()
ylim = axis.get_ylim()
print(ylim)
axis.set_ylim([ylim[0],5.3958578])

plt.show()
```
