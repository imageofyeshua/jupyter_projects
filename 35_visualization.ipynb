```python
import matplotlib.pyplot as plt
```

```python
x = 3
y = 5

plt.plot(x,y,'bs')

plt.axis('square')
plt.axis([-6,6,-6,6])
plt.grid()

plt.show()
```

```python
x = [-4,5,2,4,5,6,3]
y = [0,3,5,6,2,-3,4]

for i in range(0,len(x)):
    plt.plot(x[i],y[i],'o',label='point %s' %(1+i))

plt.axis('square')
plt.grid()
plt.legend()
plt.show()
```

```python
plt.plot(4,2,'r+')

axis = plt.gca()
ylim = axis.get_ylim()
print(ylim)
axis.set_ylim([ylim[0],5.3958578])

plt.show()
```

```python
import sympy as sym

x = sym.symbols('x')
y = x**2 - 3*x

for pnti in range(-11,12):
    plt.plot(pnti,y.subs(x,pnti),'o')

plt.xlabel('x')
plt.ylabel(r'$(fx) = x^2 - 3x$')
plt.show()
```

```python
p1 = [-3,-1]
p2 = [4,4]

plt.plot([p1[0],p2[0]],[p1[1],p2[1]],color=[.4,.2,.7],linewidth=5)

plt.axis('square')
plt.axis([-6,6,-6,6])
plt.show()
```

```python
x = 3
y = 5

plt.plot(x,y,'ro')
plt.plot([0,x],[0,y])

plt.axis('square')
plt.axis([-6,6,-6,6])
plt.grid()

plt.plot([-6,6],[0,0],'k')
plt.plot([0,0],[-6,6],'k')

plt.show()
```

```python
x = range(-20,21)

for xi in x:
    plt.plot([0,xi],[0,abs(xi)**(1/2)])

plt.show()
```

```python
plt.plot([0,2],[2,2],'r')
plt.plot([0,2],[0,0],'k')
plt.plot([0,0],[0,2],'g')
plt.plot([2,2],[0,2],'m')

plt.axis('square')
plt.axis([-3,5,-3,5])
plt.title('A Colorful Square')

plt.show()
```

```python
import numpy as np
# y = mx + b

x = [-5,5]
m = -2
b = 1

# y = [340,2345]
# for i in range(0,len(x)):
#    y[i] = m*x[i] + b

y = m*np.array(x) + b

print(x,y)

plt.plot(x,y,label='y=%sx+%s' %(m,b))
plt.axis('square')
plt.grid()
plt.xlim(x)
plt.ylim(x)

axis = plt.gca()
plt.plot(axis.get_xlim(),[0,0],'k--')
plt.plot([0,0],axis.get_xlim(),'k--')
# print(axis.get_xlim())
plt.legend()

plt.show()
```

```python
x = [-5,5]
m = [.7,-1.25]
b = [-2,.75]

for i in range(0,len(m)):
    y = m[i]*np.array(x) + b[i]
    plt.plot(x,y,label='y = %sx + %s' %(m[i],b[i]))

plt.legend()
plt.axis('square')
plt.grid()
plt.xlim(x)
plt.ylim(x)
plt.xlabel('x')
plt.ylabel('y')

axis = plt.gca()
plt.plot(axis.get_xlim(),[0,0],'k--')
plt.plot([0,0],axis.get_xlim(),'k--')
plt.show()
```
